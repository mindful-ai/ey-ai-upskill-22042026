### Zero-shot-classification

```python
classifier = pipeline("zero-shot-classification")
```

```python
classifier("This is a course about Huggingface library and it is exiciting", candidate_labels=["sports", "politics", "business", "education"],)
```

### Named Entity Recognition

```python
model = pipeline("ner", aggregation_strategy='first')
```

```python
print(model("Elon Musk founded SpaceX in the year 5-5-2017"))
```

### Using Specific Models

```python
model = pipeline("text-generation", model="HuggingFaceTB/SmolLM2-360M")
```

```python
data = model("In this AI coursem we teach you how to: ", max_length=30, num_return_sequences=2)
```

```python
len(data)
```

```python
print(data[1]['generated_text'])
```

### Image Recognition

```python
from transformers import pipeline

image_classifier = pipeline(
    task="image-classification", model="google/vit-base-patch16-224",
)
result = image_classifier(
    "https://huggingface.co/datasets/huggingface/documentation-images/resolve/main/pipeline-cat-chonk.jpeg"
)
print(result)
```

```python
import requests
import soundfile as sf
import io
from transformers import pipeline

url = "https://huggingface.co/datasets/Narsil/asr_dummy/resolve/main/mlk.flac"

transcriber = pipeline(
    task="automatic-speech-recognition", model="openai/whisper-large-v3"
)

audio_bytes = requests.get(url).content
audio, samplerate = sf.read(io.BytesIO(audio_bytes))

result = transcriber(audio)
print(result)
```
